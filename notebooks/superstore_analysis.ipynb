```python
discount_analysis = (
    df.groupby("Discount")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order ID", "nunique"),
          Quantity=("Quantity", "sum")
      )
      .reset_index()
)

discount_analysis["Profit Margin"] = (
    discount_analysis["Profit"] / discount_analysis["Sales"]
)

discount_analysis
```

```python
category_analysis=(
    df.groupby("Category")
    .agg(
        Sales=("Sales","sum"),
        Profit=("Profit","sum")
    )
    .reset_index()
)
category_analysis
```

```python
sub_category_analysis=(
    df.groupby("Sub-Category")
    .agg(
        Sales=("Sales","sum"),
        Profit=("Profit","sum")
    )
    .reset_index()
)
sub_category_analysis ["profit margin"]=(
    sub_category_analysis["Profit"]/sub_category_analysis["Sales"]
)
sub_category_analysis.sort_values("profit margin")
```

```python
discount_analysis=(
    df.groupby(["Sub-Category","Discount"])
    .agg(
        Sales=("Sales","sum"),
        Profit=("Profit","sum")
    )
    .reset_index()
)
discount_analysis["profit margin"]=(
    discount_analysis["Profit"]/discount_analysis["Sales"]
)
# discount_analysis.sort_values("profit margin")
discount_analysis.sort_values(
    ["Sub-Category", "Discount"]
)
```

```python
discount_thresholds = discount_analysis.groupby("Sub-Category").agg(
    Max_Profitable_Discount=("Discount", lambda x:
        x[discount_analysis.loc[x.index, "profit margin"] >= 0].max()
    ),
    First_Loss_Discount=("Discount", lambda x:
        x[discount_analysis.loc[x.index, "profit margin"] < 0].min()
    )
).reset_index()

discount_policy = discount_thresholds.copy()

discount_policy["Recommended_Discount"] = 0.10

discount_policy
recommended_analysis = discount_analysis[
    discount_analysis["Discount"] == 0.10
].copy()

recommended_analysis
discount_policy["Discount_Safety_Margin"] = (
    discount_policy["First_Loss_Discount"]
    - discount_policy["Recommended_Discount"]
)

discount_policy.sort_values("Discount_Safety_Margin")
```

```python
import matplotlib.pyplot as plt

plot_data = sub_category_analysis.sort_values("profit margin")

plt.figure(figsize=(10, 6))

plt.barh(
    plot_data["Sub-Category"],
    plot_data["profit margin"]
)

plt.xlabel("Profit Margin")
plt.ylabel("Sub-Category")
plt.title("Profit Margin by Sub-Category")

plt.show()
```

```python
discount_overall = df.groupby("Discount").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).reset_index()

discount_overall["Profit Margin"] = (
    discount_overall["Profit"] / discount_overall["Sales"]
)

plt.figure(figsize=(10, 6))

plt.plot(
    discount_overall["Discount"],
    discount_overall["Profit Margin"],
    marker="o"
)

plt.axhline(0, linestyle="--")

plt.xlabel("Discount")
plt.ylabel("Profit Margin")
plt.title("Discount vs Profit Margin")

plt.show()
```

```python
discount_subcategory = df.groupby(
    ["Sub-Category", "Discount"]
).agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).reset_index()

discount_subcategory["Profit Margin"] = (
    discount_subcategory["Profit"] /
    discount_subcategory["Sales"]
)

plt.figure(figsize=(12, 8))

for sub_category, group in discount_subcategory.groupby("Sub-Category"):
    plt.plot(
        group["Discount"],
        group["Profit Margin"],
        marker="o",
        label=sub_category
    )

plt.axhline(0, linestyle="--")

plt.title("Profit Margin by Discount and Sub-Category")
plt.xlabel("Discount")
plt.ylabel("Profit Margin")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
discount_kpi = (
    df[df["Discount"].isin([0, 0.1, 0.2])]
    .groupby("Discount")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
    .reset_index()
)

discount_kpi["Profit Margin"] = (
    discount_kpi["Profit"] / discount_kpi["Sales"]
)

discount_kpi["Margin Change vs 0%"] = (
    discount_kpi["Profit Margin"]
    - discount_kpi.loc[discount_kpi["Discount"] == 0, "Profit Margin"].iloc[0]
)

discount_kpi
```

```python
recommended_discount_kpi = (
    discount_analysis[discount_analysis["Discount"] == 0.1]
    .groupby("Sub-Category")
    .agg(
        Profit=("Profit", "sum"),
        Sales=("Sales", "sum")
    )
    .reset_index()
)

recommended_discount_kpi["Profit Margin"] = (
    recommended_discount_kpi["Profit"]
    / recommended_discount_kpi["Sales"]
)

profitable_categories = (
    recommended_discount_kpi["Profit Margin"] > 0
).sum()

loss_categories = (
    recommended_discount_kpi["Profit Margin"] < 0
).sum()

print("Profitable Sub-Categories:", profitable_categories)
print("Loss-Making Sub-Categories:", loss_categories)

recommended_discount_kpi
```

```python
discount_check = (
    discount_analysis[
        discount_analysis["Discount"].isin([0, 0.2])
    ]
    .pivot(
        index="Sub-Category",
        columns="Discount",
        values="profit margin"
    )
    .reset_index()
)

discount_check.columns.name = None

discount_check = discount_check.rename(
    columns={
        0.0: "Margin_0%",
        0.2: "Margin_20%"
    }
)

discount_check["Recommended_10%"] = (
    (discount_check["Margin_0%"] > 0) &
    (discount_check["Margin_20%"] > 0)
)

discount_check
```

```python
# استخراج هامش الربح عند 0% و20% لكل Sub-Category
margin_comparison = discount_analysis[
    discount_analysis["Discount"].isin([0, 0.2])
].pivot(
    index="Sub-Category",
    columns="Discount",
    values="profit margin"
).reset_index()

# إعادة تسمية الأعمدة
margin_comparison.columns = [
    "Sub-Category",
    "Margin_0%",
    "Margin_20%"
]

# ترتيب الفئات حسب هامش الربح عند 20%
chart_data = margin_comparison.sort_values(
    "Margin_20%"
).reset_index(drop=True)

# الرسم
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 8))

for _, row in chart_data.iterrows():
    ax.plot(
        [0, 1],
        [row["Margin_0%"], row["Margin_20%"]],
        marker="o"
    )

    ax.text(
        1.02,
        row["Margin_20%"],
        row["Sub-Category"],
        va="center",
        fontsize=9
    )

# خط نقطة التعادل
ax.axhline(0, linestyle="--")

ax.set_xticks([0, 1])
ax.set_xticklabels(["0% Discount", "20% Discount"])

ax.set_ylabel("Profit Margin")
ax.set_title("Profit Margin Change from 0% to 20% Discount")

ax.set_xlim(-0.05, 1.35)

plt.tight_layout()
plt.show()
```
